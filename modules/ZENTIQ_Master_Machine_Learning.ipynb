```python
# Check for unresolved placeholders

placeholder_pattern = r"\{[^}]+\}"

placeholder_count = retail_df["instruction"].str.contains(
    placeholder_pattern,
    regex=True,
    na=False
).sum()

print("PLACEHOLDER QUALITY CHECK")
print("-" * 40)
print("Instructions with placeholders:", placeholder_count)

if placeholder_count > 0:
    display(
        retail_df[
            retail_df["instruction"].str.contains(
                placeholder_pattern,
                regex=True,
                na=False
            )
        ][["instruction", "intent"]].head(10)
    )
else:
    print("No unresolved placeholders found.")
```

```python
# Identifying all unresolved placeholders.

import re
from collections import Counter

placeholders = []

for instruction in retail_df["instruction"].dropna():
    placeholders.extend(
        re.findall(r"\{([^}]+)\}", instruction)
    )

print("PLACEHOLDER SUMMARY")
print("-" * 40)

for name, count in Counter(placeholders).items():
    print(f"{name}: {count}")

print("Total placeholders:", len(placeholders))
```

## TF-IDF Text Vectorization

*Objective*
*   Convert the text questions in the Nigerian Retail Support dataset into numerical features that a machine learning model can understand.
*   TF-IDF (Term Frequency–Inverse Document Frequency) measures the importance of words and word combinations in each question. It helps the model identify meaningful terms and patterns for predicting the correct intent.


## Configure the TF-IDF Vectorizer

*Configure the vectorizer to:*
*   Convert all text to lowercase.
*   Remove common English stop words.
*   Consider single words and two-word combinations (unigrams and bigrams).
*   Limit the vocabulary to a maximum of 10,000 features.

## Transform the Dataset
* Fit the vectorizer on the training questions only, then use the same fitted vectorizer to transform the validation and testing questions.

```python
# IMPORT TF-IDF VECTORIZER

from sklearn.feature_extraction.text import TfidfVectorizer

print("TF-IDF Vectorizer imported successfully.")
```

```python
# CONFIGURE TF-IDF

tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    max_features=10000
)

print("TF-IDF configuration completed.")
```

```python
# FIT AND TRANSFORM TEXT DATA

X_train_tfidf = tfidf.fit_transform(X_train)

X_val_tfidf = tfidf.transform(X_val)

X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF transformation completed.")

print("\nTraining matrix:", X_train_tfidf.shape)
print("Validation matrix:", X_val_tfidf.shape)
print("Testing matrix:", X_test_tfidf.shape)
```

All three datasets have been transformed into numerical feature matrices, and the results are consistent:

Dataset         Shape              Meaning
Training        (3500, 2615)       3,500 questions represented by 2,615 features
Validation      (750, 2615)        750 questions represented by the same 2,615 features
Testing         (750, 2615)         750 questions represented by the same 2,615 features

*What this tells us:*
* The vectorizer identified 2,615 text features from the training questions.
* All three datasets use the sause the same feature space, so they are compatible with the same model.
* The training set was used to fit the vectorizer, preventing vocabulary leakage from the validation and test sets.

### Workflow 39: Build the Nigerian Retail Support Prediction Pipeline

### Objective

Create a function that accepts a customer question, predicts its intent and retrieves the corresponding response template.


*What this does:*

* Loads the saved TF-IDF vectorizer.

* Loads the trained Logistic Regression classifier.

* Loads the intent-to-response dictionary.

```python
### Step 1: Import the saved components

import joblib
import json

vectorizer = joblib.load(
    dataset_dir / "processed" / "zentIQ_tfidf_vectorizer.pkl"
)

loaded_model = joblib.load(
    dataset_dir / "processed" / "zentIQ_logistic_regression_model.pkl"
)

with open(
    dataset_dir / "processed" / "zentIQ_intent_response_mapping.json",
    "r",
    encoding="utf-8"
) as file:
    loaded_mapping = json.load(file)

print("All saved components loaded successfully.")
```

```python
### Step 2: Create the prediction function

def predict_customer_query(question):

    question_tfidf = vectorizer.transform([question])

    predicted_intent = loaded_model.predict(question_tfidf)[0]

    response = loaded_mapping.get(
        predicted_intent,
        "Sorry, I could not find a suitable response."
    )

    return predicted_intent, response
```

```python
from pathlib import Path
import ast
import sys

# ZENTIQ project directory
PROJECT_DIR = Path(r"C:\Users\everm\Documents_Local\ZENTIQ")

# Map Python import names to pip package names
PACKAGE_MAP = {
    "sklearn": "scikit-learn",
    "PIL": "Pillow",
    "cv2": "opencv-python",
    "yaml": "PyYAML",
    "dotenv": "python-dotenv",
    "bs4": "beautifulsoup4",
    "streamlit": "streamlit",
    "joblib": "joblib",
    "pandas": "pandas",
    "numpy": "numpy",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "scipy": "scipy",
    "nltk": "nltk",
    "spacy": "spacy",
    "transformers": "transformers",
    "torch": "torch",
    "tensorflow": "tensorflow",
    "requests": "requests",
    "flask": "flask",
    "fastapi": "fastapi",
    "sqlalchemy": "SQLAlchemy",
    "openpyxl": "openpyxl",
    "xlrd": "xlrd",
    "plotly": "plotly",
    "altair": "altair",
    "xgboost": "xgboost",
    "lightgbm": "lightgbm"
}

# Standard library modules that do not need pip installation
standard_library = set(sys.stdlib_module_names)

detected_packages = set()

# Scan Python files, excluding virtual environments and cache folders
excluded_folders = {"venv", ".venv", "env", "__pycache__", ".git"}

for python_file in PROJECT_DIR.rglob("*.py"):
    if any(part in excluded_folders for part in python_file.parts):
        continue

    try:
        tree = ast.parse(
            python_file.read_text(encoding="utf-8")
        )

        for node in ast.walk(tree):
            if isinstance(node, ast.Import):
                for alias in node.names:
                    module = alias.name.split(".")[0]

                    if module not in standard_library:
                        package = PACKAGE_MAP.get(module, module)
                        detected_packages.add(package)

            elif isinstance(node, ast.ImportFrom):
                if node.module:
                    module = node.module.split(".")[0]

                    if module not in standard_library:
                        package = PACKAGE_MAP.get(module, module)
                        detected_packages.add(package)

    except Exception as error:
        print(f"Skipped {python_file.name}: {error}")

# Include the main libraries used by the ZENTIQ project
detected_packages.update({
    "pandas",
    "numpy",
    "scikit-learn",
    "joblib",
    "streamlit"
})

# Create requirements.txt
requirements_path = PROJECT_DIR / "requirements.txt"

requirements_path.write_text(
    "\n".join(sorted(detected_packages, key=str.lower)) + "\n",
    encoding="utf-8"
)

print("SUCCESS! requirements.txt has been updated.")
print(f"\nTotal libraries detected: {len(detected_packages)}")
print("\nLibraries added:")

for package in sorted(detected_packages, key=str.lower):
    print("-", package)

print("\nSaved to:", requirements_path)
```
